> ### a) Basic try/except


````python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Error: Cannot divide by zero"

print(safe_divide(10, 2))
print(safe_divide(10, 0))
````

> ### b) Adding Error Handling to Agent Tools
>> Agents must return structured error output, not raw exceptions.

````python
def calculator(a, b, op):
    try:
        a = float(a)
        b = float(b)

        if op == '+':
            return {"ok": True, "result": a + b}
        elif op == '-':
            return {"ok": True, "result": a - b}
        elif op == '*':
            return {"ok": True, "result": a * b}
        elif op == '/':
            return {"ok": True, "result": a / b}
        else:
            return {"ok": False, "error": "Unknown operator"}

    except Exception as e:
        print(e)
        return {"ok": False, "error": str(e)}

error_result_dict = calculator(10, 0, '/')
if error_result_dict['ok']:
    print(error_result_dict['result'])
else:
    print(f"Error: {error_result_dict['error']}")
#calculator(10, 0, '/')
````

> ### c) Retrying Tool Calls (Useful for APIs & IO)

````python
def call_with_retry(func, retries=2):
    for attempt in range(retries + 1):
        try:
            return {"ok": True, "result": func()}
        except Exception as e:
            print(f"Attempt {attempt} failed:", e)
    return {"ok": False, "error": "All retries failed"}

call_with_retry(lambda: 1/0)
````

## 4. Some Important Libraries

> ### 🔵 a) What is `the` `re` Library?

> Python’s **`re`** library provides tools for working with **regular expressions (regex)** — a compact, rule-based language used to search, extract, and manipulate text.

> You use the `re` library when you need to:

> - **Search** text for a pattern  
> - **Extract** structured information from raw text  
> - **Validate** text formats (emails, numbers, tokens)  
> - **Replace** text that matches a pattern  
> - **Split** text using flexible rules  

> Regular expressions act like **mini-languages** that let you describe patterns precisely and compactly.  
They are widely used in **agent development**

>### Character Classes (What to Match)

| Pattern | Meaning                               | Example               |
|---------|----------------------------------------|------------------------|
| `\d`    | digit `[0–9]`                          | `42`                   |
| `\D`    | non-digit                              | `A`, `#`               |
| `\w`    | word char `a–z`, `A–Z`, `0–9`, `_`     | `hello_world`          |
| `\W`    | non-word character                     | `%`, `?`, `.`          |
| `\s`    | whitespace (space, tab, newline)       | `\t`, `\n`, `" "`      |
| `\S`    | non-whitespace                         | `a`, `1`, `#`          |
| `.`     | any character except newline           | `a`, `#`, `9`          |


> ### Quantifiers (How Many Times to Match)

| Pattern   | Meaning                       | Example Match       |
|-----------|-------------------------------|----------------------|
| `*`       | 0 or more                     | `a*` → `""`, `a`, `aa` |
| `+`       | 1 or more                     | `a+` → `a`, `aa`     |
| `?`       | optional (0 or 1)             | `colou?r` → color/colour |
| `{n}`     | exactly n                     | `\d{3}` → `123`      |
| `{n,}`    | at least n                    | `\d{2,}` → `23`, `5678` |
| `{n,m}`   | between n and m               | `\d{2,4}` → `12`, `1234` |
| `.*`      | any chars, greedy             | `"abc {123}"`        |
| `.*?`     | any chars, **non-greedy**     | stops at first match |


> ### Anchors (Positions in Text)

| Pattern | Meaning                                 | Example                  |
|---------|-------------------------------------------|---------------------------|
| `^`     | start of string (or line in multiline)   | `^Hello`                 |
| `$`     | end of string                             | `world$`                |
| `\b`    | word boundary                             | `\bcat\b` → cat only     |
| `\B`    | not a word boundary                       | `\Bcat` → in `concatenate` |
| `\A`    | start of string only                      | similar to `^`           |
| `\Z`    | end of string only                        | similar to `$`           |


> ### Groups and Capturing

| Pattern       | Meaning                                | Example                              |
|----------------|----------------------------------------|----------------------------------------|
| `(abc)`       | capturing group                         | extract `abc`                         |
| `(\w+)`       | capture a word                          | tool names, keys                      |
| `(?:abc)`     | non-capturing group                     | grouping without saving                |
| `(?P<name>...)` | named capturing group                 | `(?P<value>\d+)`                       |
| `(a|b)`       | OR group                                | matches `a` or `b`                    |
| `(?=...)`     | positive lookahead                      | match before pattern                  |
| `(?!...)`     | negative lookahead                      | ensure something does NOT follow      |
| `(?<=...)`    | positive lookbehind                     | match after pattern                   |
| `(?<!...)`    | negative lookbehind                     | ensure something does NOT come before |


> ### Patterns Commonly Used in Agent Development

| Purpose                       | Pattern                          |
|-------------------------------|-----------------------------------|
| Extract tool name             | `<tool:(\w+)>`                    |
| Extract JSON block            | `\{.*?\}`                         |
| Extract numbers               | `\d+`                             |
| Extract key-value             | `(\w+):\s*(.+)`                   |
| Extract Python identifiers    | `[a-zA-Z_][a-zA-Z0-9_]*`          |
| Extract arguments section     | `args:\s*(\{.*?\})`               |
| Extract bracketed content     | `\[(.*?)\]`                       |

# 📦 **Class-Based Agent State**

In real agentic systems, an agent must keep track of everything that happens during a conversation or task.
A class-based structure makes this clean, organized, and extensible.

The AgentState class encapsulates:

✔️ Messages

All user and assistant messages exchanged so far.

✔️ Observations

Outputs returned by tools (e.g., calculator, search).

✔️ Memory

Long-term information the agent stores and recalls later.

✔️ Tool Logs

A history of every tool invocation — useful for debugging and traceability.

This object acts like the “brain state” of the agent.

image.png

##🧠 Why Use a Class?:

Keeps all state in one organized place

Easy to pass around to different components

Simplifies resetting, updating, saving, or inspecting state

Aligns with how frameworks (LangChain, AutoGen, CrewAI) structure their agents internally

````python
class AgentState:
    """
    A simple class to hold all state used by an agent:
    - messages (user + assistant)
    - observations (tool outputs)
    - memory (persistent data)
    - tool logs (action history)
    """

    def __init__(self):
        self.messages = []
        self.observations = []
        self.memory = {}
        self.tool_log = []

    # ----------- MESSAGE HANDLING -----------
    def add_message(self, role, content):
        """Add a user/assistant message."""
        self.messages.append({"role": role, "content": content})

    # ----------- OBSERVATION HANDLING -----------
    def add_observation(self, tool, input_data, output):
        """Record a tool's output."""
        obs = {
            "tool": tool,
            "input": input_data,
            "output": output
        }
        self.observations.append(obs)
        self.tool_log.append(obs)  # log tool calls also

    # ----------- MEMORY HANDLING -----------
    def remember(self, key, value):
        """Store persistent memory."""
        self.memory[key] = value

    def recall(self, key, default=None):
        """Retrieve memory safely."""
        return self.memory.get(key, default)

    # ----------- STATE SUMMARY -----------
    def summary(self):
        """Return a compact state summary."""
        return {
            "messages": len(self.messages),
            "observations": len(self.observations),
            "memory_items": len(self.memory),
            "tool_calls": len(self.tool_log),
        }
````

##Lets start


````python
# Create the agent state object
state = AgentState()

# Add messages
state.add_message("user", "What is 10 + 4?")
state.add_message("assistant", "Let me calculate that.")

# Add tool observation
state.add_observation("calculator", "10 + 4", 14)

# Store memory
state.remember("preferred_language", "English")

# Retrieve memory
state.recall("preferred_language")

# Print summary
state.summary()

# View full internal state
state.__dict__
````

# ✅ Step 2: Define Simple Tools

````python
def tool_search(query):
    return f"Search results for '{query}' ... (simulated)"

def tool_calculate(expression):
    try:
        result = eval(expression)
        return f"Result: {result}"
    except:
        return "Invalid math expression"
````

# ✅ Step 3: Create a Simple Agent

Now that we have our `AgentState` and basic tools, we can build a **simple Python agent**.

The agent will:

1. **Read user input**
2. **Decide which tool to call** (rule-based)
3. **Run the tool and collect its output**
4. **Store everything in `AgentState`**
5. **Generate a response back to the user**

````python
class SimpleAgent:
    def __init__(self):
        self.state = AgentState()

    def decide_tool(self, text):
        """Rule-based: pick a tool based on simple keyword detection"""
        if "search" in text:
            return "search"
        if "calc" in text or any(c.isdigit() for c in text):
            return "calculate"
        return None

    def run(self, user_input):
        self.state.add_message("user", user_input)

        tool = self.decide_tool(user_input)

        if tool == "search":
            query = user_input.replace("search", "").strip()
            output = tool_search(query)
            self.state.add_observation("search", query, output)
            response = f"🔍 I used the search tool.\n{output}"

        elif tool == "calculate":
            expr = user_input.replace("calc", "").strip()
            output = tool_calculate(expr)
            self.state.add_observation("calculate", expr, output)
            response = f"🧮 I used the calculator.\n{output}"

        else:
            response = "I don't know which tool to use. Try 'search X' or 'calc 2+2'."

        self.state.add_message("assistant", response)
        return response
````

# ✅ Step 4: Using the Agent

````python
agent = SimpleAgent()

print(agent.run("search python tutorial"))
print(agent.run("calc 5 * 8"))
print(agent.run("hello agent"))
````

# 🧩 Task 2 — Improve Tool Detection

Right now, the agent uses simple keyword matching.

Modify decide_tool so that it:

- Uses ```text.lower()```

- Accepts more natural language commands

- Examples that should work:

```
"Can you search Python basics for me?"
"Please calculate 12 / 4"
"Flip this sentence backward"

```

````python
#@title Solution to task 2
# ============================================================
# Improved Tool Detection for Task 2
# ============================================================

def decide_tool(user_input: str):
    """
    Improved natural language tool selector.
    Uses lowercase, flexible phrasing, and multiple triggers.
    """
    text = user_input.lower()

    # -----------------------------
    # Search triggers
    # -----------------------------
    if (
        "search" in text
        or "look up" in text
        or "find" in text
        or "google" in text
        or text.startswith("can you search")
        or text.startswith("please search")
    ):
        return "search"

    # -----------------------------
    # Calculator triggers
    # -----------------------------
    if (
        "calculate" in text
        or "compute" in text
        or "what is" in text
        or any(op in text for op in ["+", "-", "*", "/", "%"])
    ):
        return "calculator"

    # -----------------------------
    # Reverse-text triggers
    # -----------------------------
    # These come from Task 1, but detection improved here
    if (
        "reverse" in text
        or "flip" in text
        or "mirror" in text
        or "backward" in text
        or "backwards" in text
        or "flip this" in text
    ):
        return "reverse"

    # Nothing matched
    return None
````
